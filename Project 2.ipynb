### Outlier Detection

Outliers are values that are unusually high or low compared to the rest of the data. 
The IQR method is used to identify potential outliers.

##### **Total Price:**

```python
df["Total Price"].plot(kind="box", color="teal")
plt.title("Total Price Outliers")
plt.ylabel("Total Price")
plt.show()
```

##### Observation

- The box plot shows several potential outliers above the upper whisker, indicating that some orders have unusually high Total Price values.

```python
Q1= df["Total Price"].quantile(0.25)
Q3 = df["Total Price"].quantile(0.75)
print("Q1: ",Q1)
print("Q3: ",Q3)
```

```python
IQR = Q3 - Q1
print("IQR: ",IQR)
```

```python
lower_bound = Q1-1.5*IQR
upper_bound = Q3 + 1.5* IQR
print("Lower Bound: ",lower_bound)
print("Upper Bound: ",upper_bound)
```

##### Checking for Outliers

Values below the lower bound and above the upper bound are considered potential outliers.

```python
lower_outliers = df[df["Total Price"]< lower_bound]
upper_outliers = df[df["Total Price"]> upper_bound]

print("Lower Outliers: ",len(lower_outliers))
print("Upper Outliers: ",len(upper_outliers))
```

##### Observation

- No lower outliers were found in Total Price.
- 8 upper outliers were identified, indicating that a small number of orders have unusually high Total Price values.

```python
upper_outliers
```

### Decide whether they are genuine or errors

The identified outliers were inspected to determine whether they were caused by data errors. The Total Price values 
are consistent with the corresponding Quantity and Unit Price values. Therefore, these observations are retained 
as valid high-value orders rather than being removed.

##### **Unit Price:**

```python
df["UnitPrice"].plot(kind="box",color="teal")
plt.title("Unit Price Outliers")
plt.ylabel("Unit Price")
plt.show()
```

```python
lower_outliers= df[df["UnitPrice"]<lower_bound]
upper_outliers = df[df["UnitPrice"]>upper_bound]
print("Lower Outliers: ",len(lower_outliers))
print("Upper Outliers: ",len(upper_outliers))
```

##### Final Observation:
